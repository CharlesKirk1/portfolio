Here is a body of starting code that calculates the cosine.  Use the function calls as examples to define correlation, euclidean and jaccard measures below.  The set up for using Pandas Series is also given.

```python
import pandas as pd
import numpy as np

def cosine(x,y):
    dpxy = (x*y).sum() # dot product of the vectors
    #print(dpxy)
    lenx = np.sqrt((x*x).sum()) # vector len = square root of sum of  squares
    leny = np.sqrt((y*y).sum())
    #print(lenx,leny)
    return dpxy/(lenx*leny)
```

```python
x = pd.Series([1,1,1,1]) # create a series out of the vector of values
y = pd.Series([2,2,2,2])
cosine(x,y)
```
