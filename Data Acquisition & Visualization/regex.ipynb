8. Display occurrences of numbers that have 5 or more digits the same, e.g. 00000 or 11111 or 22222 etc.
<p> Here is where Python alters the interpretation of a regular expression feature in its results.  If you use the notation (\d) to create a group for back referencing purposes and then use \1 to refer to it, Python will only report the last occurrence of the group. This is an example of using a while loop to keep searching for additional occurrences, accumulating them into a list.  The tweets[m.start():m.end()] is brute force getting the whole match back.

```python
#regex = re.compile(r'00000|11111|22222|etc') # this tedious approach would give results like above
regex = re.compile(r'([0-9])\1\1\1\1\1')
ml = []
m = regex.search(tweets)
while m:
    ml.append(tweets[m.start():m.end()])
    m = regex.search(tweets, m.end())
print(len(ml),ml[:10]) # almost 1600
```

9. As a final task, open the 'weather.csv' file and let's use pattern matching and substitution to partially fix this file replacing the '"T"' text with '0.001'.
