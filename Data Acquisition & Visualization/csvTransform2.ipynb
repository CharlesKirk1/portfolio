```python
def reviseData(dataRec, csvout):
    dataRec.pop(0) #idnumber can be eliminated
    
    # eliminate the date item that is a string (pop); but keep the date integers of year, month, day
    
    # when low_wind value is NA then replace with high_wind
    if dataRec[99] =="NA":
        dataRec[99] = dataRec[98] #fix these placeholder 98 and 99 indexes
    else:
        pass
    
    # where precip == "T" replace with 0.001 as we want all numbers in the column
    #if dataRec[??]  =='T':
          # ??? = 0.001
        # else:  # limit other precip values to 2 decimal places
           #  = round(float(dataRec[??]), 2)
    # convert the weather event string to a binary vector
    eventsbin = binarizeEvents(dataRec[99]) #send over the string of events; get a list back
    dataRec.pop(99) #remove this string; fix 99
    dataRec += eventsbin  # concatenate original and binarized events
            
    csvout.writerow(dataRec)
```
