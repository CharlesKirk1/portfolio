Below is an example Python function that shows several features of the Python language.  In this case we are defining a function that can be called upon to input a list of numbers until you enter the word 'end'.  The style of Python is show where statements under the control of a statement (while or for loop) are indented.  You have to Run this block to create the function but it doesn't execute the function.  Instead you have to call it as in the following block.

```python
def demoFunc():
    values = []
    i = input("Enter value or end to stop: ")
    while(i != 'end'):
        values.append(int(i)) 
        i = input("Enter value or end to stop: ")
    values.sort()
    ln = len(values)
    print('smallest=',values[0])
    print('middle=',values[(ln-1)//2])
    print('largest=',values[ln-1])
    sum = 0
    for v in values:
        sum += v
    print('sum=',sum)
    for v in values:
        print(v,'-->',v*100.0/sum)
```

```python
demoFunc() #run the function above and enter numbers; when done type 'end' without the quotes.
```

Very subtle in the demoFunc() code was the input of a number and its conversion to a number. It's seen in the line:

This is called type casting.  There are times you need to be sure the Python data type is either integer, float, string or boolean.  Enclosing the rogue data variable inside the parenthesis of int() float() str() or bool() you can change the data type.  

```python
t = '56.700'
f = float(t)
print (f)
print (int(f))
print (round(f))
u = '0'
v = int(u)
print (v)
print (bool(float(t)),bool(u),bool(v),bool(''),bool(' '), bool(None))
```

QUESTION:  Explain all the values output in the above code sequence here.

The float() functon causes t to be type cast as a float, so f is set to a float data type. This causes it to be printed with a number after the decimal point. The int(f) function causes the number after the decimal to be dropped while the round() function causes the value of f to be rounded up to 57. The int() function converts the string "0" to the number 0 so v is set to the number 0. Finally the boolean function returns True if the input is not empty (like a string with no characters), not None, and not the integer 0, and returns False otherwise.

Working with strings is a very common data preprocessing and transformational task. Write the string function code to do the transformations as suggested in the comments in as general terms as possible.  Minimize use of hard coded numbers.

```python
s = "Juniata College"
t = "Huntingdon, PA 16652-2119      "
caps = s.upper() # make the string all capital letters
city = t.split()[0][0:-1] # pull out the city of an city,-state-zip line (split)
zipcode = t.split()[-1].strip()[0:-5]  #pull out the zipcode and remove any surrounding spaces, keep only the 5 digit code
print(caps)
print(city)
print(zipcode)
```

Building up strings is also a common task.  One of the Python 3 features is the .format() function applied to a format string.   Let's go the other way and build up strings for either printing or output to a file. For the variables given, create a format string to match the 911 address pattern as shown in the comment.

```python
addrnumber = 1700
street= "Moore St."
city = "Huntingdon"
state = "Pa" # want to be sure state is all caps
zipcode = 6652 #it's a bad idea for zips to be integer.  New England zips need leading zeros. Ensure that!
addrstring = "{0} {1},{2}, {3} {4:05d}".format(addrnumber, street, city, state.upper(), zipcode)# add in the other variables too
# resulting addrstring should have the pattern including the two commas and potential leading zeroes
# 1700 Moore St.,Huntingdon, PA 06652
print(addrstring)
```
